```python
class Node:

    def __init__(self, board, parent=None, direction=None, depth=0):
        self.board = board  # 3x3 list of lists
        self.parent = parent
        self.direction = direction  # Direction taken to reach this state
        self.depth = depth  # Current depth of the node


    def find_blank(self):
        for i in range(3):
            for j in range(3):
                if self.board[i][j] == 0:
                    return i, j

    def get_neighbors(self):
        neighbors = []
        r, c = self.find_blank()

        # Moves: (direction, row_offset, col_offset)
        moves = [
            ("Right", 0, 1),
            ("Left", 0, -1),
            ("Down", 1, 0),
            ("Up", -1, 0),
        ]

        for direction, dr, dc in moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board = [row[:] for row in self.board]
                new_board[r][c], new_board[nr][nc] = (
                    new_board[nr][nc],
                    new_board[r][c],
                )
                neighbors.append(
                    Node(new_board, self, direction, self.depth + 1)
                )
        return neighbors


def print_board(board):
    """Helper function to print the 3x3 grid cleanly."""
    for row in board:
        print(" ".join(str(val) if val != 0 else "_" for val in row))
    print()


def depth_limited_search(initial_state, goal_state, limit):
    """Performs a stack-based DFS up to a strict depth limit."""
    def to_tuple(b):
        return tuple(tuple(row) for row in b)

    root = Node(initial_state)
    stack = [root]

    # Tracks the shallowest depth we have seen a state to avoid cycles
    visited = {to_tuple(initial_state): 0}
    cutoff_occurred = False

    while stack:
        current_node = stack.pop()

        if current_node.board == goal_state:
            return current_node  # Solution found!

        if current_node.depth < limit:
            for neighbor in current_node.get_neighbors():
                board_tuple = to_tuple(neighbor.board)

                # Only explore if unseen or reached via a shorter path
                if (
                    board_tuple not in visited
                    or neighbor.depth < visited[board_tuple]
                ):
                    visited[board_tuple] = neighbor.depth
                    stack.append(neighbor)
        else:
            cutoff_occurred = True

    return "cutoff" if cutoff_occurred else "failure"


def iterative_deepening_search(initial_state, goal_state):
    """Loops through increasing depths to find the optimal path."""
    depth = 0
    while True:
        result = depth_limited_search(initial_state, goal_state, depth)

        if result != "cutoff":
            if isinstance(result, Node):
                # Backtrack to build the node path from start to goal
                path = []
                curr = result
                while curr:
                    path.append(curr)
                    curr = curr.parent
                path.reverse()

                print(f"--- IDDFS Solution Path (Found at Depth {depth}) ---")
                for node in path:
                    if node.direction:
                        print(f"Action: Moved {node.direction}")
                    else:
                        print("Action: Initial State")
                    print_board(node.board)
                return True
            else:
                print("No solution found.")
                return False

        depth += 1


# Example Usage:
initial = [[2,8,3],[1,6,4],[7,0,5]]

goal =[[1,2,3],[8,0,4],[7,6,5]]

iterative_deepening_search(initial, goal)
```
