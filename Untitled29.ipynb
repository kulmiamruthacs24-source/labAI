```python
class Node:

    def __init__(self, board, parent=None, direction=None):
        self.board = board  # 3x3 list of lists
        self.parent = parent
        self.direction = direction  # Direction taken to reach this state


    def find_blank(self):
        for i in range(3):
            for j in range(3):
                if self.board[i][j] == 0:
                    return i, j

    def get_neighbors(self):
        neighbors = []
        r, c = self.find_blank()

        # Moves: (direction, row_offset, col_offset)
        moves = [
            ("Right", 0, 1),
            ("Left", 0, -1),
            ("Down", 1, 0),
            ("Up", -1, 0),
        ]

        for direction, dr, dc in moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board = [row[:] for row in self.board]
                new_board[r][c], new_board[nr][nc] = (
                    new_board[nr][nc],
                    new_board[r][c],
                )
                neighbors.append(Node(new_board, self, direction))
        return neighbors


def print_board(board):
    """Helper function to print the 3x3 grid cleanly."""
    for row in board:
        print(" ".join(str(val) if val != 0 else "_" for val in row))
    print()


def dfs(initial_state, goal_state):
    def to_tuple(b):
        return tuple(tuple(row) for row in b)

    root = Node(initial_state)
    stack = [root]
    visited = set()
    visited.add(to_tuple(initial_state))

    while stack:
        current_node = stack.pop()

        # If goal is found, reconstruct and print ONLY the winning path
        if current_node.board == goal_state:
            # Backtrack to build the node path
            path = []
            while current_node:
                path.append(current_node)
                current_node = current_node.parent
            path.reverse()  # Reverse to read from start to goal

            print("--- Solution Path Found ---")
            for node in path:
                if node.direction:
                    print(f"Action: Moved {node.direction}")
                else:
                    print("Action: Initial State")
                print_board(node.board)

            print("--- Goal State Reached! ---")
            return True

        # Explore neighbors silently
        for neighbor in current_node.get_neighbors():
            b_tuple = to_tuple(neighbor.board)
            if b_tuple not in visited:
                visited.add(b_tuple)
                stack.append(neighbor)

    print("--- No Solution Found ---")
    return False


# Example Usage:
initial = [[2,8,3],[1,6,4],[7,0,5]]

goal =[[1,2,3],[8,0,4],[7,6,5]]

dfs(initial, goal)
```

```python
class Node:

    def __init__(self, board, parent=None, direction=None):
        self.board = board  # 3x3 list of lists
        self.parent = parent
        self.direction = direction  # Direction taken to reach this state

    def find_blank(self):
        for i in range(3):
            for j in range(3):
                if self.board[i][j] == 0:
                    return i, j

    def get_neighbors(self):
        neighbors = []
        r, c = self.find_blank()

        # Moves: (direction, row_offset, col_offset)
        moves = [
            ("Right", 0, 1),
            ("Left", 0, -1),
            ("Down", 1, 0),
            ("Up", -1, 0),
        ]

        for direction, dr, dc in moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board = [row[:] for row in self.board]
                new_board[r][c], new_board[nr][nc] = (
                    new_board[nr][nc],
                    new_board[r][c],
                )
                neighbors.append(Node(new_board, self, direction))
        return neighbors


def print_board(board):
    """Helper function to print the 3x3 grid cleanly."""
    for row in board:
        # Replaces 0 with a blank space for better visualization
        print(" ".join(str(val) if val != 0 else "_" for val in row))
    print()


def dfs(initial_state, goal_state):
    def to_tuple(b):
        return tuple(tuple(row) for row in b)

    root = Node(initial_state)
    stack = [root]
    visited = set()
    visited.add(to_tuple(initial_state))

    step = 1
    print("--- Starting Traversal ---")

    while stack:
        current_node = stack.pop()

        # Print current step details
        print(f"Step {step}:")
        if current_node.direction:
            print(f"Action: Moved {current_node.direction}")
        else:
            print("Action: Initial State")

        print_board(current_node.board)
        step += 1

        # Check if we reached the goal
        if current_node.board == goal_state:
            print("--- Goal State Reached! ---")
            return True

        # Explore neighbors
        for neighbor in current_node.get_neighbors():
            b_tuple = to_tuple(neighbor.board)
            if b_tuple not in visited:
                visited.add(b_tuple)
                stack.append(neighbor)

    print("--- Traversal Finished (No Solution) ---")
    return False


# Example Usage:
initial = [[2, 8, 3], [1, 6, 4], [7, 0, 5]]

goal = [[1, 2, 3], [8, 0, 4], [7, 6, 5]]

dfs(initial, goal)
```
