```python
def get_user_boolean(prompt_msg):
    """Safely gets a true/false input from the user."""
    while True:
        user_input = input(prompt_msg).strip().lower()
        if user_input in ['true', 't', '1']:
            return True
        elif user_input in ['false', 'f', '0']:
            return False
        print("Invalid input. Please enter 'true' or 'false'.")

def generate_user_defined_table():
    user_rows = []

    print("--- STEP 1: ENTER VALUES FOR ALL 8 ROWS ---")
    # Collect all 8 rows of input from the user
    for i in range(1, 9):
        print(f"\n[Row {i}/8]")
        A = get_user_boolean(f"  Enter value for A: ")
        B = get_user_boolean(f"  Enter value for B: ")
        C = get_user_boolean(f"  Enter value for C: ")
        user_rows.append((A, B, C))

    print("\n--- STEP 2: GENERATING FINAL OUTPUT TABLE ---")
    # Print the table header matching the image layout
    header = f"{'A':<7} | {'B':<7} | {'C':<7} || {'A ∨ C':<7} | {'B ∨ ¬C':<7} || {'KB':<7} | {'α':<7}"
    print(header)
    print("-" * len(header))

    # Process and print each user-inputted row
    for A, B, C in user_rows:
        # Perform the remaining logical operations automatically
        A_or_C = A or C
        B_or_notC = B or (not C)
        KB = A_or_C and B_or_notC
        alpha = A or B

        # Format strings to lowercase strings to match image style
        a_str = str(A).lower()
        b_str = str(B).lower()
        c_str = str(C).lower()
        a_c_str = str(A_or_C).lower()
        b_nc_str = str(B_or_notC).lower()
        kb_str = str(KB).lower()
        alpha_str = str(alpha).lower()

        # Display the formatted row
        print(f"{a_str:<7} | {b_str:<7} | {c_str:<7} || {a_c_str:<7} | {b_nc_str:<7} || {kb_str:<7} | {alpha_str:<7}")
    print("-" * len(header))

if __name__ == "__main__":
    generate_user_defined_table()
```
