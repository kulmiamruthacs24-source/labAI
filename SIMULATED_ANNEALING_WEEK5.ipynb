```python
import math
import random

def get_conflicts(board):
    """
    Calculates the number of mutually attacking pairs of queens.
    The index represents the column, and the value represents the row.
    """
    conflicts = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row conflict
            if board[i] == board[j]:
                conflicts += 1
            # Diagonal conflicts: check if row distance equals column distance
            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbor(board):
    """
    Generates a neighbor by randomly selecting one queen
    and moving it to a different row in its column.
    """
    neighbor = list(board)
    col = random.randint(0, 7)
    current_row = board[col]

    # Pick a new row that is different from the current row
    new_row = random.randint(0, 7)
    while new_row == current_row:
        new_row = random.randint(0, 7)

    neighbor[col] = new_row
    return neighbor

def print_board(board):
    """Helper function to print the board visually."""
    for row in range(8):
        line = ""
        for col in range(8):
            if board[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

def simulated_annealing_8queens(initial_temp=10.0, cooling_rate=0.995, min_temp=0.001):
    # Initialize board randomly: each element is a row index (0-7) for that column
    current_solution = [random.randint(0, 7) for _ in range(8)]
    current_energy = get_conflicts(current_solution)

    # Track historic absolute best found during the walk
    best_solution = list(current_solution)
    best_energy = current_energy

    temperature = initial_temp
    step = 0

    # Loop until the system freezes or a perfect solution is found
    while temperature > min_temp and current_energy > 0:
        neighbor = get_neighbor(current_solution)
        neighbor_energy = get_conflicts(neighbor)

        delta_energy = neighbor_energy - current_energy

        # If the neighbor reduces conflicts, always accept it
        if delta_energy < 0:
            current_solution = neighbor
            current_energy = neighbor_energy
            if neighbor_energy < best_energy:
                best_solution = list(neighbor)
                best_energy = neighbor_energy
        else:
            # If the neighbor is worse, accept it with a temperature-dependent probability
            acceptance_probability = math.exp(-delta_energy / temperature)
            if random.random() < acceptance_probability:
                current_solution = neighbor
                current_energy = neighbor_energy

        # Cool down the system smoothly
        temperature *= cooling_rate
        step += 1

    return best_solution, best_energy, step

# --- Execution ---
if __name__ == "__main__":
    random.seed() # Seed with system clock for true randomness

    print("Searching for an 8-Queens solution...")
    best_board, conflicts, total_steps = simulated_annealing_8queens()

    print(f"\nTerminated after {total_steps} temperature steps.")
    print(f"Final conflict count: {conflicts}")
    print(f"Board array layout (Columns 0-7): {best_board}\n")

    print("Visual Chessboard Layout:")
    print_board(best_board)

    if conflicts == 0:
        print("\n🎉 Success! A valid 8-Queens configuration was found.")
    else:
        print("\n❌ Trapped in a local optimum. Try running the algorithm again.")
```
