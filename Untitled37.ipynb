```python
import re

def is_variable(expr):
    """
    Returns True if the expression is a variable.
    By convention, single lowercase letters (x, y, z, etc.) are variables.
    """
    return isinstance(expr, str) and expr.islower() and len(expr) == 1

def is_constant(expr):
    """
    Returns True if the expression is a constant.
    Constants are strings that are not variables (e.g., numbers, 'A', 'B', 'John').
    """
    return isinstance(expr, str) and not is_variable(expr) and not isinstance(expr, list)

def occurs_check(var, expr):
    """
    Step 1(b)a & 1(c)a: Check if a variable occurs inside another expression.
    """
    if var == expr:
        return True
    if isinstance(expr, list):
        # Search inside arguments (skipping the predicate/function name at index 0)
        return any(occurs_check(var, arg) for arg in expr[1:])
    return False

def apply_substitution(subst, expr):
    """
    Applies a substitution dictionary to an expression recursively.
    """
    if not subst:
        return expr
    if is_variable(expr):
        return subst.get(expr, expr)
    if isinstance(expr, list):
        return [expr[0]] + [apply_substitution(subst, arg) for arg in expr[1:]]
    return expr

def unify(psi1, psi2):
    """
    The main Unification Algorithm following the slide steps.
    """
    # Step 1: If psi1 or psi2 is a variable or constant
    if (is_variable(psi1) or is_constant(psi1)) or (is_variable(psi2) or is_constant(psi2)):
        # a) If they are identical, return {} (NIL / empty substitution)
        if psi1 == psi2:
            return {}

        # b) Else if psi1 is a variable
        elif is_variable(psi1):
            if occurs_check(psi1, psi2):
                return "FAILURE"
            else:
                return {psi1: psi2}  # returns { psi2 / psi1 }

        # c) Else if psi2 is a variable
        elif is_variable(psi2):
            if occurs_check(psi2, psi1):
                return "FAILURE"
            else:
                return {psi2: psi1}  # returns { psi1 / psi2 }

        # d) Else return FAILURE
        else:
            return "FAILURE"

    # Step 2: If the initial Predicate symbol in psi1 and psi2 are not same
    if psi1[0] != psi2[0]:
        return "FAILURE"

    # Step 3: If psi1 and psi2 have a different number of arguments
    if len(psi1) != len(psi2):
        return "FAILURE"

    # Step 4: Set Substitution set (SUBST) to NIL (empty dict)
    subst = {}

    # Step 5: For i=1 to the number of elements in psi1
    for i in range(1, len(psi1)):
        # a) Call Unify with the ith element of psi1 and psi2
        s = unify(psi1[i], psi2[i])

        # b) If S = failure then returns Failure
        if s == "FAILURE":
            return "FAILURE"

        # c) If S is not NIL (not empty)
        if s:
            # a. Apply S to the remainder of both terms
            for j in range(i + 1, len(psi1)):
                psi1[j] = apply_substitution(s, psi1[j])
                psi2[j] = apply_substitution(s, psi2[j])

            # b. SUBST = APPEND(S, SUBST)
            new_subst = {}
            for k, v in subst.items():
                new_subst[k] = apply_substitution(s, v)
            new_subst.update(s)
            subst = new_subst

    # Step 6: Return SUBST
    return subst

def parse_expression(s):
    """
    Robust stack-based parser to reliably convert strings like 'P(x, f(y))'
    into structured lists like ['P', 'x', ['f', 'y']].
    """
    s = s.replace(" ", "")
    if '(' not in s:
        return s

    tokens = re.split(r'([(),])', s)
    tokens = [t for t in tokens if t]

    stack = []
    current = []

    for token in tokens:
        if token == '(':
            pred = current.pop()
            stack.append(current)
            current = [pred]
        elif token == ')':
            if stack:
                completed = current
                current = stack.pop()
                current.append(completed)
        elif token == ',':
            continue
        else:
            current.append(token)

    return current[0] if current else s

def format_expression(expr):
    """
    Helper to turn list format back into string format for display
    """
    if isinstance(expr, list):
        return f"{expr[0]}({', '.join(format_expression(arg) for arg in expr[1:])})"
    return str(expr)

# --- Main User Interaction ---
if __name__ == "__main__":
    print("=== Unification Algorithm ===")
    print("Rules: Lowercase single letters (x, y, z) are variables. Everything else is a constant/predicate.")
    print("Example input: P(x, f(y)) and P(A, f(B))\n")

    # Accept inputs from the user
    exp1_str = input("Enter Expression 1: ")
    exp2_str = input("Enter Expression 2: ")

    # Parse expressions
    expr1 = parse_expression(exp1_str)
    expr2 = parse_expression(exp2_str)

    # Run the unification logic
    result = unify(expr1, expr2)

    print("\n----------------------------------")
    if result == "FAILURE":
        print("Output: The entered expressions are NOT unified.")
    else:
        print("Output: The entered expressions are UNIFIED!")
        if not result:
            print("Substitution Set: NIL (Expressions were already identical)")
            print(f"Unified Expression: {format_expression(expr1)}")
        else:
            print("Substitution Set (Mating):")
            for var, val in result.items():
                print(f"  {format_expression(val)} / {var}")

            # Show final unified expression by applying the substitution mappings
            final_expr = apply_substitution(result, parse_expression(exp1_str))
            print(f"Unified Expression: {format_expression(final_expr)}")
    print("----------------------------------")
```
