```python
class VacuumAgent:
    def __init__(self, location, status_a, status_b):
        self.location = location
        self.room_status = {
            'A': status_a.strip().capitalize(),
            'B': status_b.strip().capitalize()
        }

    def sense(self):
        return self.room_status[self.location]

    def act(self, status):
        if status == 'Dirty':
            print(f"Location {self.location} is Dirty. Action: Suck dirt.")
            self.room_status[self.location] = 'Clean'
        else:
            print(f"Location {self.location} is Clean. Action: No operation needed.")

        # Move to the other room if needed
        if self.location == 'A':
            print("Action: Move to Location B.")
            self.location = 'B'
        else:
            print("Action: Move to Location A.")
            self.location = 'A'

def main():
    # User inputs
    loc = input("Enter starting location of vacuum (A or B): ").strip().upper()
    stat_a = input("Enter status of Room A (Dirty or Clean): ")
    stat_b = input("Enter status of Room B (Dirty or Clean): ")

    # Initialize agent
    vacuum = VacuumAgent(loc, stat_a, stat_b)

    print("\n--- Starting Vacuum Cleaner Simulation ---")

    # Run simulation for both rooms
    for step in range(2):
        print(f"\nStep {step + 1}:")
        current_status = vacuum.sense()
        vacuum.act(current_status)

    print("\n--- Simulation Complete. Both rooms are checked/cleaned. ---")

if __name__ == "__main__":
    main()
```
