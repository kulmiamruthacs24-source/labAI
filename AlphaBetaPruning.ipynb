```python


def alpha_beta_search(state):
    value, action = max_value(state, float('-inf'), float('inf'))
    return value, action


def max_value(state, alpha, beta):
    if isinstance(state, int):
        return state, None

    v = float('-inf')
    best_action = None

    for i, child in enumerate(state):
        child_value, _ = min_value(child, alpha, beta)

        if child_value > v:
            v = child_value
            best_action = i + 1

        if v >= beta:
            return v, best_action

        alpha = max(alpha, v)

    return v, best_action


def min_value(state, alpha, beta):
    if isinstance(state, int):
        return state, None

    v = float('inf')
    best_action = None

    for i, child in enumerate(state):
        child_value, _ = max_value(child, alpha, beta)

        if child_value < v:
            v = child_value
            best_action = i + 1

        if v <= alpha:
            return v, best_action

        beta = min(beta, v)

    return v, best_action


def create_tree():
    choice = int(input("Enter 1 for Terminal Node, 0 for Non-Terminal Node: "))

    if choice == 1:
        return int(input("Enter utility value: "))

    n = int(input("Enter number of child nodes: "))
    children = []

    for i in range(n):
        print("Enter child", i + 1)
        children.append(create_tree())

    return children


# Main Program
print("ALPHA-BETA SEARCH")

print("Enter the game tree:")
tree = create_tree()

value, action = alpha_beta_search(tree)

print("\nOptimal Utility Value:", value)

if action is not None:
    print("Best Action: Child", action)
else:
    print("Root is a terminal node.")
```
