```python
import heapq

# Define the goal state for the 8-puzzle
GOAL_STATE = (1, 2, 3,
              8, 0, 4,
              7, 6, 5)  # 0 represents the empty/blank space

class Node:
    def __init__(self, state, parent=None, action=None, g=0, h=0):
        self.state = state      # Tuple representing the 3x3 grid configuration
        self.parent = parent    # Link to the parent node
        self.action = action    # Action taken to reach this state
        self.g = g              # Cost from initial state to current node
        self.h = h              # Heuristic cost (estimated cost to goal)
        self.f = g + h          # Total estimated cost (f = g + h)

    # Define comparison operator for priority queue ordering (ascending f value)
    def __lt__(self, other):
        return self.f < other.f

def get_misplaced_tiles(state):
    """Calculates the number of tiles that are not in their goal positions."""
    misplaced = 0
    for i in range(9):
        # Count only non-zero tiles that do not match the goal state position
        if state[i] != 0 and state[i] != GOAL_STATE[i]:
            misplaced += 1
    return misplaced

def get_successors(node):
    """Generates valid child nodes by moving the empty space (0) Up, Down, Left, or Right."""
    successors = []
    state = node.state
    blank_idx = state.index(0)
    row, col = blank_idx // 3, blank_idx % 3

    # Moves mapped to (row_change, col_change, action_name)
    moves = {
        (-1, 0): 'Up',
        (1, 0): 'Down',
        (0, -1): 'Left',
        (0, 1): 'Right'
    }

    for (dr, dc), action in moves.items():
        new_row, new_col = row + dr, col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank_idx = new_row * 3 + new_col
            # Swap the blank space with the target tile
            new_state = list(state)
            new_state[blank_idx], new_state[new_blank_idx] = new_state[new_blank_idx], new_state[blank_idx]
            new_state_tuple = tuple(new_state)

            # Create child node
            g_cost = node.g + 1
            h_cost = get_misplaced_tiles(new_state_tuple)
            child = Node(state=new_state_tuple, parent=node, action=action, g=g_cost, h=h_cost)
            successors.append(child)

    return successors

def a_star_search(initial_state):
    """Executes the A* Search algorithm to find the shortest path to the goal state."""
    # Initialize the start node
    start_node = Node(state=initial_state, g=0, h=get_misplaced_tiles(initial_state))

    # Frontier as a priority queue
    frontier = []
    heapq.heappush(frontier, start_node)

    # Set to keep track of visited states and their best g costs
    explored = {initial_state: 0}

    while frontier:
        # Pop the node with the lowest f value
        current_node = heapq.heappop(frontier)

        # Goal test
        if current_node.state == GOAL_STATE:
            return reconstruct_path(current_node)

        # Generate actions/successors
        for child in get_successors(current_node):
            # If the state hasn't been visited or we found a cheaper path to it
            if child.state not in explored or child.g < explored[child.state]:
                explored[child.state] = child.g
                heapq.heappush(frontier, child)

    return None  # Failure (No solution found)

def reconstruct_path(node):
    """Traces back from the goal node to the start node to collect the sequence of steps."""
    path = []
    curr = node
    while curr.parent is not None:
        path.append((curr.action, curr.state))
        curr = curr.parent
    path.reverse()
    return path

def print_puzzle(state):
    """Helper function to print the 3x3 puzzle layout grid."""
    for i in range(0, 9, 3):
        print(f"[{state[i]} {state[i+1]} {state[i+2]}]")
    print()

# --- Example Execution ---
if __name__ == "__main__":
    # 0 represents the blank spot
    initial_puzzle = (2, 8, 3,
                      1, 6, 4,
                      7, 0, 5)

    print("Initial State:")
    print_puzzle(initial_puzzle)

    solution = a_star_search(initial_puzzle)

    if solution:
        print(f"Solution found in {len(solution)} moves!\n")
        for step, (action, state) in enumerate(solution, 1):
            print(f"Step {step}: Move {action}")
            print_puzzle(state)
    else:
        print("No solution exists for this configuration.")
```
