```python
import math

# Initialize the 3x3 board with empty spaces
board = [' ' for _ in range(9)]

def print_board():
    """Prints the current game board to the console."""
    print("\n")
    for i in range(3):
        row = f" {board[i*3]} | {board[i*3+1]} | {board[i*3+2]} "
        print(row)
        if i < 2:
            print("---+---+---")
    print("\n")

def check_winner(b, player):
    """Checks if the given player ('X' or 'O') has won the game."""
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8], # Horizontal rows
        [0, 3, 6], [1, 4, 7], [2, 5, 8], # Vertical columns
        [0, 4, 8], [2, 4, 6]             # Diagonals
    ]
    return any(all(b[cell] == player for cell in combo) for combo in win_conditions)

def is_board_full(b):
    """Returns True if there are no empty spaces left on the board."""
    return ' ' not in b

def minimax(b, depth, is_maximizing):
    """
    Minimax algorithm to find the optimal move for the AI.
    AI is 'O' (maximizing), Human is 'X' (minimizing).
    """
    # Base cases: Check for terminal states (win/loss/draw)
    if check_winner(b, 'O'):
        return 1
    if check_winner(b, 'X'):
        return -1
    if is_board_full(b):
        return 0

    if is_maximizing:
        best_score = -math.inf
        for i in range(9):
            if b[i] == ' ':
                b[i] = 'O'
                score = minimax(b, depth + 1, False)
                b[i] = ' ' # Undo move
                best_score = max(score, best_score)
        return best_score
    else:
        best_score = math.inf
        for i in range(9):
            if b[i] == ' ':
                b[i] = 'X'
                score = minimax(b, depth + 1, True)
                b[i] = ' ' # Undo move
                best_score = min(score, best_score)
        return best_score

def get_ai_move():
    """Calculates the best possible move for the AI ('O')."""
    best_score = -math.inf
    best_move = None
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            score = minimax(board, 0, False)
            board[i] = ' ' # Undo move
            if score > best_score:
                best_score = score
                best_move = i
    return best_move

def play_game():
    """Main game loop."""
    print("Welcome to Tic-Tac-Toe AI!")
    print("Positions are numbered 1 through 9 like this:")
    print(" 1 | 2 | 3 \n---+---+---\n 4 | 5 | 6 \n---+---+---\n 7 | 8 | 9 ")
    print("You are 'X' and the AI is 'O'.")

    print_board()

    while True:
        # --- Human Turn ---
        while True:
            try:
                move = int(input("Enter your move (1-9): ")) - 1
                if 0 <= move <= 8 and board[move] == ' ':
                    board[move] = 'X'
                    break
                else:
                    print("Invalid move! The spot is either taken or out of bounds.")
            except ValueError:
                print("Please enter a valid number between 1 and 9.")

        print_board()

        if check_winner(board, 'X'):
            print("Congratulations! You won!")
            break
        if is_board_full(board):
            print("It's a draw!")
            break

        # --- AI Turn ---
        print("AI is making a move...")
        ai_move = get_ai_move()
        if ai_move is not None:
            board[ai_move] = 'O'

        print_board()

        if check_winner(board, 'O'):
            print("AI wins! Better luck next time.")
            break
        if is_board_full(board):
            print("It's a draw!")
            break

if __name__ == "__main__":
    play_game()
```
