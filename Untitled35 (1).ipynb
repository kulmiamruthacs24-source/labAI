```python
import random

def get_conflicts(state):
    """Calculates the total number of attacking pairs of queens on the board."""
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    """Finds the best neighbor move and records exactly which queen moved."""
    current_conflicts = get_conflicts(state)
    best_state = list(state)
    best_conflicts = current_conflicts
    moved_col = None
    old_row = None
    new_row = None
    n = len(state)

    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbor_conflicts = get_conflicts(neighbor)

                if neighbor_conflicts < best_conflicts:
                    best_conflicts = neighbor_conflicts
                    best_state = neighbor
                    moved_col = col
                    old_row = state[col]
                    new_row = row

    return best_state, best_conflicts, moved_col, old_row, new_row

def print_board(state):
    """Prints a visual 4x4 layout of the board."""
    for row in range(4):
        line = ""
        for col in range(4):
            if state[col] == row:
                line += "[Q] "
            else:
                line += "[. ] "
        print(line.strip())
    print()

def hill_climbing_with_movement():
    # Generate random configuration
    current_state = [random.randint(0, 3) for _ in range(4)]
    current_conflicts = get_conflicts(current_state)

    print("--- Initial State Configuration ---")
    print(f"State Array: {current_state} | Conflicts: {current_conflicts}\n")
    print_board(current_state)

    step = 1
    while True:
        neighbor_state, neighbor_conflicts, col, old_r, new_r = get_best_neighbor(current_state)

        if neighbor_conflicts >= current_conflicts:
            # Hit a peak or plateau
            break

        print(f"👉 Step {step}: Moved Queen in Column {col} from Row {old_r} to Row {new_r}")
        print(f"New Array: {neighbor_state} | Conflicts remaining: {neighbor_conflicts}\n")

        current_state = neighbor_state
        current_conflicts = neighbor_conflicts
        print_board(current_state)
        step += 1

    return current_state, current_conflicts

# --- Run until a perfect solution is reached ---
if __name__ == "__main__":
    attempt = 1
    while True:
        print(f"=================== RUN ATTEMPT #{attempt} ===================")
        final_state, final_conflicts = hill_climbing_with_movement()

        if final_conflicts == 0:
            print("🎉 Success! Found a solution with zero conflicts.")
            break
        else:
            print("❌ Stuck in a local optimum (cannot make any better moves). Restarting...\n")
            attempt += 1
```
